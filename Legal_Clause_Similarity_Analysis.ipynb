```python
"""
Legal Clause Similarity Analysis
Deep Learning Assignment 2 - NUCES FAST Islamabad

This module implements two baseline architectures for identifying semantic similarity
between legal clauses without using pre-trained transformers.
"""

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from typing import List, Tuple, Dict
import random
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

# Deep Learning Libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, precision_recall_curve,
                             auc, confusion_matrix, classification_report)

print("TensorFlow version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))

# ============================================================================
# 1. DATA LOADING AND PREPROCESSING
# ============================================================================

class LegalClauseDataLoader:
    """Handles loading and initial processing of legal clause dataset"""

    def __init__(self, data_path: str = None):
        self.data_path = data_path
        self.clauses_by_category = {}
        self.all_clauses = []

    def download_dataset(self):
        """Download dataset from Kaggle using kagglehub"""
        print("Downloading dataset from Kaggle...")
        try:
            import kagglehub
        except ImportError:
            print("Installing kagglehub...")
            os.system('pip install -q kagglehub')
            import kagglehub

        # Download dataset
        path = kagglehub.dataset_download("bahushruth/legalclausedataset")
        self.data_path = path
        print(f"Dataset downloaded successfully to: {path}")

    def load_data(self) -> pd.DataFrame:
        """Load all CSV files from the dataset"""
        if self.data_path is None:
            self.download_dataset()

        csv_files = [f for f in os.listdir(self.data_path) if f.endswith('.csv')]
        print(f"Found {len(csv_files)} category files")

        all_data = []
        for file in csv_files:
            category = file.replace('.csv', '')
            try:
                df = pd.read_csv(os.path.join(self.data_path, file))
                df['category'] = category
                all_data.append(df)
                self.clauses_by_category[category] = df
                print(f"Loaded {len(df)} clauses from {category}")
            except Exception as e:
                print(f"Error loading {file}: {e}")

        combined_df = pd.concat(all_data, ignore_index=True)
        self.all_clauses = combined_df
        return combined_df


class ClausePairGenerator:
    """Generate positive and negative clause pairs for training"""

    def __init__(self, df: pd.DataFrame, text_column: str = 'clause_text'):
        self.df = df
        self.text_column = text_column

    def create_pairs(self, n_positive: int = 5000, n_negative: int = 5000) -> pd.DataFrame:
        """
        Create pairs of clauses with similarity labels
        Positive pairs: same category
        Negative pairs: different categories
        """
        positive_pairs = []
        negative_pairs = []

        # Generate positive pairs (same category)
        categories = self.df['category'].unique()
        for category in categories:
            cat_clauses = self.df[self.df['category'] == category][self.text_column].tolist()
            if len(cat_clauses) < 2:
                continue

            n_pairs = min(n_positive // len(categories), len(cat_clauses) * (len(cat_clauses) - 1) // 2)

            for _ in range(n_pairs):
                idx1, idx2 = random.sample(range(len(cat_clauses)), 2)
                positive_pairs.append({
                    'clause1': cat_clauses[idx1],
                    'clause2': cat_clauses[idx2],
                    'label': 1,
                    'category': category
                })

        # Generate negative pairs (different categories)
        for _ in range(n_negative):
            cat1, cat2 = random.sample(list(categories), 2)
            clause1 = random.choice(self.df[self.df['category'] == cat1][self.text_column].tolist())
            clause2 = random.choice(self.df[self.df['category'] == cat2][self.text_column].tolist())
            negative_pairs.append({
                'clause1': clause1,
                'clause2': clause2,
                'label': 0,
                'category': f"{cat1}_{cat2}"
            })

        # Combine and shuffle
        all_pairs = positive_pairs + negative_pairs
        random.shuffle(all_pairs)

        pairs_df = pd.DataFrame(all_pairs)
        print(f"\nGenerated {len(positive_pairs)} positive pairs and {len(negative_pairs)} negative pairs")
        print(f"Total pairs: {len(pairs_df)}")
        print(f"Class distribution:\n{pairs_df['label'].value_counts()}")

        return pairs_df


class TextPreprocessor:
    """Preprocess and tokenize text data"""

    def __init__(self, max_words: int = 10000, max_len: int = 200):
        self.max_words = max_words
        self.max_len = max_len
        self.tokenizer = None

    def fit_tokenizer(self, texts: List[str]):
        """Fit tokenizer on text data"""
        self.tokenizer = Tokenizer(num_words=self.max_words, oov_token='<OOV>')
        self.tokenizer.fit_on_texts(texts)
        print(f"Vocabulary size: {len(self.tokenizer.word_index)}")

    def texts_to_sequences(self, texts: List[str]) -> np.ndarray:
        """Convert texts to padded sequences"""
        sequences = self.tokenizer.texts_to_sequences(texts)
        padded = pad_sequences(sequences, maxlen=self.max_len, padding='post', truncating='post')
        return padded

    def prepare_pair_data(self, pairs_df: pd.DataFrame) -> Tuple:
        """Prepare clause pairs for model input"""
        all_texts = pairs_df['clause1'].tolist() + pairs_df['clause2'].tolist()
        self.fit_tokenizer(all_texts)

        clause1_seq = self.texts_to_sequences(pairs_df['clause1'].tolist())
        clause2_seq = self.texts_to_sequences(pairs_df['clause2'].tolist())
        labels = pairs_df['label'].values

        return clause1_seq, clause2_seq, labels


# ============================================================================
# 2. MODEL ARCHITECTURES
# ============================================================================

class BiLSTMSiameseNetwork:
    """
    Baseline 1: Bidirectional LSTM-based Siamese Network
    Uses BiLSTM to encode clauses and compute similarity
    """

    def __init__(self, vocab_size: int, embedding_dim: int = 128,
                 lstm_units: int = 64, max_len: int = 200):
        self.vocab_size = vocab_size
        self.embedding_dim = embedding_dim
        self.lstm_units = lstm_units
        self.max_len = max_len
        self.model = None
        self.history = None

    def build_model(self):
        """Build BiLSTM Siamese architecture"""
        # Shared embedding layer
        embedding_layer = layers.Embedding(
            input_dim=self.vocab_size,
            output_dim=self.embedding_dim,
            input_length=self.max_len,
            mask_zero=True
        )

        # Shared BiLSTM encoder
        def create_encoder():
            inputs = layers.Input(shape=(self.max_len,))
            x = embedding_layer(inputs)
            x = layers.Bidirectional(layers.LSTM(self.lstm_units, return_sequences=True))(x)
            x = layers.Bidirectional(layers.LSTM(self.lstm_units // 2))(x)
            x = layers.Dense(128, activation='relu')(x)
            x = layers.Dropout(0.3)(x)
            return Model(inputs=inputs, outputs=x)

        # Create Siamese network
        encoder = create_encoder()

        input1 = layers.Input(shape=(self.max_len,))
        input2 = layers.Input(shape=(self.max_len,))

        encoded1 = encoder(input1)
        encoded2 = encoder(input2)

        # Compute similarity features
        diff = layers.Subtract()([encoded1, encoded2])
        mult = layers.Multiply()([encoded1, encoded2])
        concat = layers.Concatenate()([encoded1, encoded2, diff, mult])

        # Classification head
        x = layers.Dense(64, activation='relu')(concat)
        x = layers.Dropout(0.3)(x)
        x = layers.Dense(32, activation='relu')(x)
        output = layers.Dense(1, activation='sigmoid')(x)

        self.model = Model(inputs=[input1, input2], outputs=output)
        self.model.compile(
            optimizer='adam',
            loss='binary_crossentropy',
            metrics=['accuracy', tf.keras.metrics.AUC(name='auc')]
        )

        return self.model

    def train(self, X1_train, X2_train, y_train, X1_val, X2_val, y_val,
              epochs: int = 50, batch_size: int = 32):
        """Train the model"""
        callbacks = [
            EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6)
        ]

        self.history = self.model.fit(
            [X1_train, X2_train], y_train,
            validation_data=([X1_val, X2_val], y_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )

        return self.history

    def get_summary(self):
        """Get model summary"""
        return self.model.summary()


class AttentionBasedEncoder:
    """
    Baseline 2: Attention-based Encoder
    Uses self-attention mechanism to capture clause semantics
    """

    def __init__(self, vocab_size: int, embedding_dim: int = 128,
                 attention_heads: int = 4, max_len: int = 200):
        self.vocab_size = vocab_size
        self.embedding_dim = embedding_dim
        self.attention_heads = attention_heads
        self.max_len = max_len
        self.model = None
        self.history = None

    def build_model(self):
        """Build Attention-based architecture"""
        # Shared embedding
        embedding_layer = layers.Embedding(
            input_dim=self.vocab_size,
            output_dim=self.embedding_dim,
            input_length=self.max_len,
            mask_zero=True
        )

        def create_encoder():
            inputs = layers.Input(shape=(self.max_len,))
            x = embedding_layer(inputs)

            # Multi-head self-attention
            attention_output = layers.MultiHeadAttention(
                num_heads=self.attention_heads,
                key_dim=self.embedding_dim // self.attention_heads
            )(x, x)

            # Add & Norm
            x = layers.Add()([x, attention_output])
            x = layers.LayerNormalization()(x)

            # Feed-forward network
            ff = layers.Dense(128, activation='relu')(x)
            ff = layers.Dense(self.embedding_dim)(ff)

            # Add & Norm
            x = layers.Add()([x, ff])
            x = layers.LayerNormalization()(x)

            # Global average pooling
            x = layers.GlobalAveragePooling1D()(x)
            x = layers.Dense(128, activation='relu')(x)
            x = layers.Dropout(0.3)(x)

            return Model(inputs=inputs, outputs=x)

        # Create Siamese network
        encoder = create_encoder()

        input1 = layers.Input(shape=(self.max_len,))
        input2 = layers.Input(shape=(self.max_len,))

        encoded1 = encoder(input1)
        encoded2 = encoder(input2)

        # Similarity computation
        diff = layers.Subtract()([encoded1, encoded2])
        mult = layers.Multiply()([encoded1, encoded2])
        concat = layers.Concatenate()([encoded1, encoded2, diff, mult])

        # Classification
        x = layers.Dense(64, activation='relu')(concat)
        x = layers.Dropout(0.3)(x)
        x = layers.Dense(32, activation='relu')(x)
        output = layers.Dense(1, activation='sigmoid')(x)

        self.model = Model(inputs=[input1, input2], outputs=output)
        self.model.compile(
            optimizer='adam',
            loss='binary_crossentropy',
            metrics=['accuracy', tf.keras.metrics.AUC(name='auc')]
        )

        return self.model

    def train(self, X1_train, X2_train, y_train, X1_val, X2_val, y_val,
              epochs: int = 50, batch_size: int = 32):
        """Train the model"""
        callbacks = [
            EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6)
        ]

        self.history = self.model.fit(
            [X1_train, X2_train], y_train,
            validation_data=([X1_val, X2_val], y_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )

        return self.history

    def get_summary(self):
        """Get model summary"""
        return self.model.summary()


# ============================================================================
# 3. EVALUATION AND VISUALIZATION
# ============================================================================

class ModelEvaluator:
    """Comprehensive model evaluation and visualization"""

    def __init__(self, model, model_name: str):
        self.model = model
        self.model_name = model_name
        self.metrics = {}

    def evaluate(self, X1_test, X2_test, y_test) -> Dict:
        """Compute all evaluation metrics"""
        # Predictions
        y_pred_proba = self.model.predict([X1_test, X2_test]).flatten()
        y_pred = (y_pred_proba > 0.5).astype(int)

        # Metrics
        self.metrics = {
            'accuracy': accuracy_score(y_test, y_pred),
            'precision': precision_score(y_test, y_pred),
            'recall': recall_score(y_test, y_pred),
            'f1_score': f1_score(y_test, y_pred),
            'roc_auc': roc_auc_score(y_test, y_pred_proba),
        }

        # PR-AUC
        precision_vals, recall_vals, _ = precision_recall_curve(y_test, y_pred_proba)
        self.metrics['pr_auc'] = auc(recall_vals, precision_vals)

        # Confusion matrix
        self.metrics['confusion_matrix'] = confusion_matrix(y_test, y_pred)

        print(f"\n{self.model_name} Evaluation Results:")
        print("=" * 50)
        for metric, value in self.metrics.items():
            if metric != 'confusion_matrix':
                print(f"{metric.upper()}: {value:.4f}")

        print("\nClassification Report:")
        print(classification_report(y_test, y_pred, target_names=['Different', 'Similar']))

        return self.metrics

    def plot_training_history(self, history):
        """Plot training curves"""
        fig, axes = plt.subplots(1, 3, figsize=(18, 5))

        # Loss
        axes[0].plot(history.history['loss'], label='Train Loss')
        axes[0].plot(history.history['val_loss'], label='Val Loss')
        axes[0].set_title(f'{self.model_name} - Loss')
        axes[0].set_xlabel('Epoch')
        axes[0].set_ylabel('Loss')
        axes[0].legend()
        axes[0].grid(True)

        # Accuracy
        axes[1].plot(history.history['accuracy'], label='Train Accuracy')
        axes[1].plot(history.history['val_accuracy'], label='Val Accuracy')
        axes[1].set_title(f'{self.model_name} - Accuracy')
        axes[1].set_xlabel('Epoch')
        axes[1].set_ylabel('Accuracy')
        axes[1].legend()
        axes[1].grid(True)

        # AUC
        axes[2].plot(history.history['auc'], label='Train AUC')
        axes[2].plot(history.history['val_auc'], label='Val AUC')
        axes[2].set_title(f'{self.model_name} - AUC')
        axes[2].set_xlabel('Epoch')
        axes[2].set_ylabel('AUC')
        axes[2].legend()
        axes[2].grid(True)

        plt.tight_layout()
        plt.savefig(f'{self.model_name}_training_curves.png', dpi=300, bbox_inches='tight')
        plt.show()

    def plot_confusion_matrix(self):
        """Plot confusion matrix"""
        plt.figure(figsize=(8, 6))
        sns.heatmap(self.metrics['confusion_matrix'], annot=True, fmt='d',
                   cmap='Blues', xticklabels=['Different', 'Similar'],
                   yticklabels=['Different', 'Similar'])
        plt.title(f'{self.model_name} - Confusion Matrix')
        plt.ylabel('True Label')
        plt.xlabel('Predicted Label')
        plt.savefig(f'{self.model_name}_confusion_matrix.png', dpi=300, bbox_inches='tight')
        plt.show()


class ResultsAnalyzer:
    """Analyze and display qualitative results"""

    def __init__(self, model, preprocessor, pairs_df):
        self.model = model
        self.preprocessor = preprocessor
        self.pairs_df = pairs_df

    def show_predictions(self, X1, X2, n_samples: int = 10):
        """Show sample predictions"""
        predictions = self.model.predict([X1, X2]).flatten()

        print("\n" + "="*80)
        print("SAMPLE PREDICTIONS")
        print("="*80)

        for i in range(min(n_samples, len(predictions))):
            pred_label = "SIMILAR" if predictions[i] > 0.5 else "DIFFERENT"
            true_label = "SIMILAR" if self.pairs_df.iloc[i]['label'] == 1 else "DIFFERENT"
            confidence = predictions[i] if predictions[i] > 0.5 else 1 - predictions[i]

            print(f"\nPair {i+1}:")
            print(f"Clause 1: {self.pairs_df.iloc[i]['clause1'][:100]}...")
            print(f"Clause 2: {self.pairs_df.iloc[i]['clause2'][:100]}...")
            print(f"True Label: {true_label} | Predicted: {pred_label} (Confidence: {confidence:.3f})")
            print("-" * 80)


# ============================================================================
# 4. MAIN PIPELINE
# ============================================================================

class LegalClauseSimilarityPipeline:
    """Main pipeline orchestrating the entire workflow"""

    def __init__(self):
        self.loader = None
        self.preprocessor = None
        self.pairs_df = None
        self.models = {}
        self.results = {}

    def run(self):
        """Execute complete pipeline"""
        print("="*80)
        print("LEGAL CLAUSE SIMILARITY ANALYSIS PIPELINE")
        print("="*80)

        # Step 1: Load Data
        print("\n[1/7] Loading Dataset...")
        self.loader = LegalClauseDataLoader()
        df = self.loader.load_data()

        # Step 2: Generate Pairs
        print("\n[2/7] Generating Clause Pairs...")
        pair_generator = ClausePairGenerator(df)
        self.pairs_df = pair_generator.create_pairs(n_positive=5000, n_negative=5000)

        # Step 3: Preprocess
        print("\n[3/7] Preprocessing Text...")
        self.preprocessor = TextPreprocessor(max_words=10000, max_len=200)
        X1, X2, y = self.preprocessor.prepare_pair_data(self.pairs_df)

        # Step 4: Split Data
        print("\n[4/7] Splitting Data...")
        indices = np.arange(len(y))
        train_idx, temp_idx = train_test_split(indices, test_size=0.3, random_state=42, stratify=y)
        val_idx, test_idx = train_test_split(temp_idx, test_size=0.5, random_state=42, stratify=y[temp_idx])

        X1_train, X2_train, y_train = X1[train_idx], X2[train_idx], y[train_idx]
        X1_val, X2_val, y_val = X1[val_idx], X2[val_idx], y[val_idx]
        X1_test, X2_test, y_test = X1[test_idx], X2[test_idx], y[test_idx]

        print(f"Train: {len(y_train)}, Val: {len(y_val)}, Test: {len(y_test)}")

        # Step 5: Train BiLSTM Model
        print("\n[5/7] Training BiLSTM Siamese Network...")
        vocab_size = min(len(self.preprocessor.tokenizer.word_index) + 1, 10000)

        bilstm_model = BiLSTMSiameseNetwork(vocab_size=vocab_size)
        bilstm_model.build_model()
        print("\nBiLSTM Model Architecture:")
        bilstm_model.get_summary()

        bilstm_history = bilstm_model.train(
            X1_train, X2_train, y_train,
            X1_val, X2_val, y_val,
            epochs=30, batch_size=32
        )

        self.models['BiLSTM'] = bilstm_model

        # Step 6: Train Attention Model
        print("\n[6/7] Training Attention-Based Encoder...")
        attention_model = AttentionBasedEncoder(vocab_size=vocab_size)
        attention_model.build_model()
        print("\nAttention Model Architecture:")
        attention_model.get_summary()

        attention_history = attention_model.train(
            X1_train, X2_train, y_train,
            X1_val, X2_val, y_val,
            epochs=30, batch_size=32
        )

        self.models['Attention'] = attention_model

        # Step 7: Evaluate Both Models
        print("\n[7/7] Evaluating Models...")

        # BiLSTM Evaluation
        bilstm_evaluator = ModelEvaluator(bilstm_model.model, "BiLSTM")
        bilstm_metrics = bilstm_evaluator.evaluate(X1_test, X2_test, y_test)
        bilstm_evaluator.plot_training_history(bilstm_history)
        bilstm_evaluator.plot_confusion_matrix()
        self.results['BiLSTM'] = bilstm_metrics

        # Attention Evaluation
        attention_evaluator = ModelEvaluator(attention_model.model, "Attention")
        attention_metrics = attention_evaluator.evaluate(X1_test, X2_test, y_test)
        attention_evaluator.plot_training_history(attention_history)
        attention_evaluator.plot_confusion_matrix()
        self.results['Attention'] = attention_metrics

        # Comparative Analysis
        self.compare_models()

        # Qualitative Results
        print("\n" + "="*80)
        print("QUALITATIVE ANALYSIS - BiLSTM MODEL")
        analyzer = ResultsAnalyzer(bilstm_model.model, self.preprocessor,
                                   self.pairs_df.iloc[test_idx].reset_index(drop=True))
        analyzer.show_predictions(X1_test, X2_test, n_samples=5)

        print("\n" + "="*80)
        print("Pipeline Complete!")
        print("="*80)

    def compare_models(self):
        """Compare performance of both models"""
        print("\n" + "="*80)
        print("COMPARATIVE ANALYSIS")
        print("="*80)

        comparison_df = pd.DataFrame(self.results).T
        print("\n", comparison_df[['accuracy', 'precision', 'recall', 'f1_score', 'roc_auc', 'pr_auc']])

        # Visualization
        fig, axes = plt.subplots(1, 2, figsize=(15, 5))

        metrics_to_plot = ['accuracy', 'precision', 'recall', 'f1_score']
        x = np.arange(len(metrics_to_plot))
        width = 0.35

        bilstm_vals = [self.results['BiLSTM'][m] for m in metrics_to_plot]
        attention_vals = [self.results['Attention'][m] for m in metrics_to_plot]

        axes[0].bar(x - width/2, bilstm_vals, width, label='BiLSTM', alpha=0.8)
        axes[0].bar(x + width/2, attention_vals, width, label='Attention', alpha=0.8)
        axes[0].set_xlabel('Metrics')
        axes[0].set_ylabel('Score')
        axes[0].set_title('Model Comparison - Classification Metrics')
        axes[0].set_xticks(x)
        axes[0].set_xticklabels(metrics_to_plot, rotation=45)
        axes[0].legend()
        axes[0].grid(True, alpha=0.3)

        auc_metrics = ['roc_auc', 'pr_auc']
        x_auc = np.arange(len(auc_metrics))

        bilstm_auc = [self.results['BiLSTM'][m] for m in auc_metrics]
        attention_auc = [self.results['Attention'][m] for m in auc_metrics]

        axes[1].bar(x_auc - width/2, bilstm_auc, width, label='BiLSTM', alpha=0.8)
        axes[1].bar(x_auc + width/2, attention_auc, width, label='Attention', alpha=0.8)
        axes[1].set_xlabel('Metrics')
        axes[1].set_ylabel('Score')
        axes[1].set_title('Model Comparison - AUC Metrics')
        axes[1].set_xticks(x_auc)
        axes[1].set_xticklabels(auc_metrics)
        axes[1].legend()
        axes[1].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.savefig('model_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()


# ============================================================================
# 5. EXECUTION
# ============================================================================

if __name__ == "__main__":
    # Set random seeds for reproducibility
    np.random.seed(42)
    tf.random.set_seed(42)
    random.seed(42)

    # Run pipeline
    pipeline = LegalClauseSimilarityPipeline()
    pipeline.run()

    print("\nAll visualizations and results have been saved!")
    print("Check the generated PNG files for training curves and comparisons")
```
